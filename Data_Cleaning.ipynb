# Data Load and Cleaning
- Load data
- Remove unnecessary column
- Data Type change
- Duplicate
- Null Value
- Save cleaned data

```python
# Imports

import numpy as np
import pandas as pd
```

```python
# Load Data

data = pd.read_csv(r"C:\Users\Hp\Desktop\BIA\DATA SCIENCE PROJECTS\ds_project_1\CAPSTONE PROJECT\Data\Fraud_Analysis_Dataset.csv")
data.head()
```

```python
# Create back-up
df = data.copy()
```

```python
# Meta-data
print("No. of rows", df.shape[0])
print("No. of columns", df.shape[1])
all_columns = df.columns.to_list()
num_columns = df.select_dtypes(include = "number").columns.to_list()
cat_columns = df.select_dtypes(include = "object").columns.to_list()
print("All columns: \n", all_columns)
print("\n Numeric Columns: \n", num_columns)
print("\n Categorical Columns: \n", cat_columns)
print ("\n Data Type: \n", df.dtypes)
print("\n Info: \n")
df.info()
```

```python
# Duplicate 
print("Duplicate count: ", df.duplicated().sum())
print("Duplicate Columns: \n")
df[df.duplicated()]
```

```python
# Null value check
df.isnull().sum()
```

```python
# Target distribution

print("Target counts:\n", df["isFraud"].value_counts())
print("\nTarget distribution (%):\n", (df["isFraud"].value_counts(normalize=True) * 100).round(2))
```

```python
# Remove unnecessary columns

df.drop(columns=["nameOrig", "nameDest"], inplace=True)
df.head(2)
```

```python
# Final structure check

df.info()
```

```python
# Save cleaned data

file_name = "cleaned_fraud_data.csv"
df.to_csv(file_name, index=False)
print(f"Cleaned data saved as: {file_name}")
```
